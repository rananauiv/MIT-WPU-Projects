```python
# Step 11: Display passenger names
print("\nPassenger Names:")
print(df["Name"].head())

# Step 12: Count passengers by Class
print("\nPassengers by Gender:")
print(df["Pclass"].value_counts())

# Step 13: Calculate average age
print("\nAverage Age:")
print(df["Age"].mean())

# Step 14: Find the highest fare
print("\nHighest Fare:")
print(df["Fare"].max())

# Step 15: Find the average fare
print("\nAverage Fare:")
print(df["Fare"].mean())


df = df.drop(['Cabin', 'Embarked'], axis=1)
df=df.drop(columns=('Cabin', 'Embarked'))
```

```python
# Step 16: Filter passengers older than 60
print("\nPassengers Older Than 60:")
print(df[df["Age"] > 60])

```

```python
# Step 17: Display passengers who survived
print("\nPassengers Who Survived:")
print(df[df["Survived"] == 1][["Name", "Sex", "Age"]].head())
```

```python
# Step 18: Count survivors and non-survivors
print("\nSurvival Counts:")
print(df["Survived"].value_counts())
```

```python
# Step 19: Calculate survival rate by gender
print("\nSurvival Rate by Gender:")
print(df.groupby("Sex")["Survived"].mean())
```

```python
# Step 20: Sort passengers by fare
print("\nTop 10 Highest Fares:")
print(
    df.sort_values("Fare", ascending=False)
      [["Name", "Fare"]]
      .head(10)
)

```

```python
# Step 21: Count passengers by passenger class
print("\nPassengers by Class:")
print(df["Pclass"].value_counts().sort_index())



```

```python
# Step 22: Calculate average age by gender
print("\nAverage Age by Gender:")
print(df.groupby("Sex")["Age"].mean())


```

```python
# Step 23: Display embarkation port counts
print("\nPassengers by Embarkation Port:")
print(df["Embarked"].value_counts())
```

```python

# Step 24: Save a copy of the dataset
df.to_csv("titanic_copy.csv", index=False)
```

```python

# 3. Remove Duplicates
df = df.drop_duplicates()
print("Duplicate Rows Removed")
```

```python
 2. Handle Missing Values
for col in df.columns:
    if df[col].isnull().any():
        if pd.api.types.is_numeric_dtype(df[col]):
            df[col] = df[col].fillna(df[col].mean())
        else:
            df[col] = df[col].fillna(df[col].mode()[0])

print("\nMissing Values Handled")
```
