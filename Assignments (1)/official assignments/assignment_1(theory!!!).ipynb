```python
from datetime import date

# 1. Dictionary: assign 5 elements, add, delete and display

my_dict = {
    "name": "ABC",
    "panel": "B",
    "rollno": 34,
    "marks": 85,
    "city": "Mumbai"
}

print()
print("1. Original Dictionary:", my_dict)

new_key = input("Enter new student information key: ")
new_value = input("Enter new student information value: ")
my_dict[new_key] = new_value

delete_key = input("Enter key to delete: ")

if delete_key in my_dict:
    del my_dict[delete_key]

print("Updated Student Information:", my_dict)


# 2. Convert two lists into a dictionary

list1 = ["name", "panel", "rollno"]
list2 = ["ABC", "B", 34]

my_dict2 = {list1[i]: list2[i] for i in range(len(list1))}

print()
print("2. Dictionary from Lists:", my_dict2)


# 3. Sort elements of a dictionary using sorted()

my_dict3 = {
    "name": "ABC",
    "panel": "B",
    "rollno": 34,
    "marks": [65, 87, 67, 94]
}

sorted_dict = dict(sorted(my_dict3.items()))

print()
print("3. Sorted Dictionary:", sorted_dict)


# 4. Convert dictionary to lists

keys_list = list(my_dict3.keys())
values_list = list(my_dict3.values())

print()
print("4. Keys List:", keys_list)
print("Values List:", values_list)


# 5. Find mean of values of dictionary

marks_dict = {
    "marks1": 23,
    "marks2": 123,
    "marks3": 43,
    "marks4": 13,
    "marks5": 39
}

mean = sum(marks_dict.values()) / len(marks_dict)

print()
print("5. Mean of Dictionary Values:", mean)


# 6. Dictionary operations

my_dict6 = {
    "name": ["Yash", "Neel", "Dev"],
    "rollno": [11, 12, 13],
    "marks": [78, 56, 98]
}

print()
print("6a. Name as Dev:")
index = my_dict6["name"].index("Dev")
print(my_dict6["name"][index])

print()
print("6b. Roll Number 12:")
index = my_dict6["rollno"].index(12)
print(my_dict6["rollno"][index])

print()
print("6c. Greatest Marks:")
print(max(my_dict6["marks"]))


# 7. Calculate frequency of letters in a string

print()
string = input("Enter a string: ")
frequency = {}

for char in string:
    if char != " ":
        frequency[char] = frequency.get(char, 0) + 1

print("Letter Frequency:", frequency)


# 8. Input numerator and denominator and output quotient and remainder as a tuple

print()
numerator = int(input("Enter numerator: "))
denominator = int(input("Enter denominator: "))

result = (numerator // denominator, numerator % denominator)

print("Quotient and Remainder Tuple:", result)


# 9. Create two date tuples and find number of days between them

print()
d1 = int(input("Enter first date day: "))
m1 = int(input("Enter first date month: "))
y1 = int(input("Enter first date year: "))

print()
d2 = int(input("Enter second date day: "))
m2 = int(input("Enter second date month: "))
y2 = int(input("Enter second date year: "))

date_tuple1 = (d1, m1, y1)
date_tuple2 = (d2, m2, y2)

date1 = date(y1, m1, d1)
date2 = date(y2, m2, d2)

days_between = abs((date2 - date1).days)

print()
print("First Date Tuple:", date_tuple1)
print("Second Date Tuple:", date_tuple2)
print("Number of Days Between Dates:", days_between)
```
